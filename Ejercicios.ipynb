**Ejercicio 1 — Inventario de la tienda**

Una tienda pequeña tiene el siguiente inventario:

Productos y precios (en pesos colombianos):
- Cuaderno: 3500
- Lápiz: 800
- Café: 6200.50
Cantidad a comprar: 3 cuadernos, 5 lápices
La tienda tiene actualmente descuentos activos: no

 TODO: declare nombre_producto (str) con el nombre de un producto de su elección

 TODO: declare precio_cuaderno (int) y precio_lapiz (int)

 TODO: declare precio_cafe (float)

 TODO: declare hay_descuento (bool) indicando si la tienda tiene descuentos activos

 TODO: declare cantidad_cuadernos (int) y cantidad_lapices (int)

TODO: verifique con type() el tipo de cada una de las variables anteriores
Requisitos:

Declare las siete variables indicadas, con el tipo de dato correcto para cada una (int, float, str, bool).
Verifique con type() que cada variable quedó con el tipo esperado.

```python
# Ejercicio 1 — Inventario de la tienda
nombre_producto = "Cuaderno"       # str
precio_cuaderno = 3500             # int
precio_lapiz = 800                 # int
precio_cafe = 6200.50              # float
hay_descuento = False              # bool
cantidad_cuadernos = 3             # int
cantidad_lapices = 5               # int

# Verificación de tipos
print("Tipo de nombre_producto:", type(nombre_producto))
print("Tipo de precio_cuaderno:", type(precio_cuaderno))
print("Tipo de precio_lapiz:", type(precio_lapiz))
print("Tipo de precio_cafe:", type(precio_cafe))
print("Tipo de hay_descuento:", type(hay_descuento))
print("Tipo de cantidad_cuadernos:", type(cantidad_cuadernos))
print("Tipo de cantidad_lapices:", type(cantidad_lapices))
```
