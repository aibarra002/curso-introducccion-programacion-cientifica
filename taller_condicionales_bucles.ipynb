Ejercicio 6 — Recorrer el inventario completo
La tienda tiene ahora un inventario con varios productos:

cantidades_en_stock = [0, 3, 12, 0, 7, 2]

```python
#  Declarar la lista de cantidades de inventario
cantidades_en_stock = [0, 3, 12, 0, 7, 2]

for indice in range(len(cantidades_en_stock)):

    #  Obtener la cantidad de la posición actual
    cantidad = cantidades_en_stock[indice]

    #  Clasificar la cantidad
    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    # 5. Mostrar el mensaje formateado
    print(f"Producto {indice}: {nivel}")
```

Ejercicio 7 — Promedio semanal de temperaturas
La semana de temperaturas quedó registrada así:

temperaturas_semana = [18.5, 21.0, 19.8, 24.3, 17.2]

```python
# 1. Declarar la lista de temperaturas
temperaturas_semana = [18.5, 21.0, 19.8, 24.3, 17.2]

# 2. Declarar acumulador y contador
suma_temperaturas = 0.0
dias_calurosos = 0

# 3. Recorrer la lista por índice
for i in range(len(temperaturas_semana)):

    # Extraer la temperatura del día actual
    temp_actual = temperaturas_semana[i]

    # 4. Sumar la temperatura al acumulador
    suma_temperaturas += temp_actual

    # 5. Evaluar si el día es caluroso e incrementar el contador
    if temp_actual > 20:
        dias_calurosos += 1

# 6. Calcular el promedio fuera del bucle
promedio_semanal = suma_temperaturas / len(temperaturas_semana)

# 7. Mostrar los resultados
print("Promedio semanal de temperatura:", promedio_semanal)
print("Cantidad de días calurosos:", dias_calurosos)
```

Ejercicio 8 — Comprar tornillos mientras alcance el presupuesto
La ferretería tiene un presupuesto para reponer tornillos:

presupuesto_ferreteria = 3000
precio_tornillo = 250

```python
# 1. Declarar presupuesto y precio
presupuesto_ferreteria = 3000
precio_tornillo = 250

# 2. Declarar el contador de compras
tornillos_comprados = 0

# 3. Bucle while: repite mientras alcance el presupuesto
while presupuesto_ferreteria >= precio_tornillo:

    # Descontar el precio del presupuesto
    presupuesto_ferreteria -= precio_tornillo

    # Incrementar el contador de tornillos
    tornillos_comprados += 1

# 4. Mostrar los resultados después del bucle
print("Tornillos comprados:", tornillos_comprados)
print("Presupuesto sobrante:", presupuesto_ferreteria)
```

Ejercicio 9 — Buscar la primera herramienta cara
La ferretería tiene los siguientes precios de herramientas:

precios_herramientas = [12000.0, 8500.0, 25000.0, 9800.0]

```python
# Declarar la lista de precios
precios_herramientas = [12000.0, 8500.0, 25000.0, 9800.0]

for precio in precios_herramientas:

    #  Evaluar si el precio supera los 20000
    if precio > 20000:
        print(f"Primera herramienta cara encontrada con precio: {precio}")
        break
         # 4. Cortar el bucle de inmediato
    else:
        print(f"Sigue revisando, precio actual: {precio}")
```

Ejercicio 10 — Revisión completa del inventario de la ferretería
La ferretería tiene el siguiente inventario, con nombre y cantidad de cada herramienta en la misma posición de cada lista:

nombres_herramientas = ["Martillo", "Taladro", "Sierra", "Destornillador", "Nivel", "Llave inglesa"]
cantidades_herramientas = [5, 0, 2, 8, 0, 1]

```python
# 1. Declarar las listas de inventario
nombres_herramientas = ["Martillo", "Taladro", "Sierra", "Destornillador", "Nivel", "Llave inglesa"]
cantidades_herramientas = [5, 0, 2, 8, 0, 1]

# 2. Recorrer ambas listas
for i in range(len(nombres_herramientas)):

    # Extraer el nombre y la cantidad de la posición actual
    nombre = nombres_herramientas[i]
    cantidad = cantidades_herramientas[i]

    # 3. Clasificar la cantidad y usar continue si está agotado
    if cantidad == 0:
        print(f"{nombre}: Agotado. Saltando a la siguiente herramienta...")
        continue  # Corta esta iteración y pasa a la siguiente

    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    # 4. Mostrar el mensaje solo para las herramientas que sí tienen stock
    print(f"{nombre}: {nivel} ({cantidad} unidades)")
```
