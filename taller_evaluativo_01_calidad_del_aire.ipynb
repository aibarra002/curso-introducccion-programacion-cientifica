Ejercicio 2
— Conversión de unidades y error frente al equipo patrón
El sensor de campo reportó una lectura y, en el mismo instante, el equipo patrón del laboratorio midió el valor de referencia:

lectura_sensor_mg = 0.0418   # en mg/m3, como lo entrega el sensor
valor_patron_ug = 40.0       # en ug/m3, medido por el equipo de referencia

```python
# Variables iniciales
lectura_sensor_mg = 0.0418
valor_patron_ug = 40.0

# 1. Conversión de la lectura del sensor a ug/m3
lectura_sensor_ug = lectura_sensor_mg * factor_conversion

# 2. Cálculo del error absoluto
error_absoluto = lectura_sensor_ug - valor_patron_ug

# 3. Cálculo del error relativo porcentual
error_relativo_porcentual = (error_absoluto / valor_patron_ug) * 100

# 4. Cálculo del error cuadrático
error_cuadratico = error_absoluto ** 2

# Mostrar resultados
print("Lectura en ug/m3:", lectura_sensor_ug)
print("Error absoluto:", error_absoluto)
print("Error relativo porcentual (%):", error_relativo_porcentual)
print("Error cuadrático:", error_cuadratico)
```

Ejercicio 3 — Cobertura temporal del lote descargado
La estación toma una lectura cada cinco minutos, es decir 12 lecturas por hora. Con las numero_lecturas del Ejercicio 1, calcule cuánto tiempo cubre el lote descargado.

```python
# 1. Declarar lecturas por hora
lecturas_por_hora = 12

# 2. Horas completas y lecturas sobrantes
horas_completas = numero_lecturas // lecturas_por_hora
lecturas_sobrantes = numero_lecturas % lecturas_por_hora

# 3. Días completos y horas sobrantes (usando 24 horas por día)
dias_completos = horas_completas // 24
horas_sobrantes = horas_completas % 24

# Mostrar resultados
print("Horas completas:", horas_completas)
print("Lecturas sobrantes:", lecturas_sobrantes)
print("Días completos:", dias_completos)
print("Horas sobrantes:", horas_sobrantes)
```

Ejercicio 4
— Validar una lectura sin decidir todavía
Antes de clasificar una medición hay que saber si es utilizable. Una lectura es válida cuando no es el centinela de dato faltante y cae dentro del rango físico que el sensor puede reportar.

lectura = 41.8

limite_superior_sensor = 500.0

```python
# 1. Declaración de variables iniciales (float)
lectura = 41.8
codigo_dato_faltante = -999.0
limite_superior_sensor = 500.0

# 2. Evaluar si es un dato faltante
es_dato_faltante = lectura == codigo_dato_faltante

# 3. Evaluar si está en el rango físico
esta_en_rango = (lectura >= 0) and (lectura <= limite_superior_sensor)

# 4. Evaluar si la lectura es válida
lectura_valida = esta_en_rango and not es_dato_faltante

# Mostrar los resultados booleanos
print("¿Es dato faltante?:", es_dato_faltante)
print("¿Está en rango?:", esta_en_rango)
print("¿Es una lectura válida?:", lectura_valida)
```

Ejercicio 6
— Limpiar la serie horaria
Estas son doce lecturas consecutivas descargadas de la estación, con dos fallas de sensor entre ellas:

```python
# 1. Definición de  lista
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]
lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

# 2. Recorrido de la lista
for lectura in lecturas_pm25:
    if lectura == -999.0:
        lecturas_descartadas += 1
        continue  # Salta el resto del cuerpo

    lecturas_validas.append(lectura)
    suma_pm25 += lectura

# 3. Cálculo del promedio sobre datos válidos
promedio_pm25 = suma_pm25 / len(lecturas_validas)

# Mostrar resultados
print("Cantidad de lecturas válidas:", len(lecturas_validas))
print("Cantidad de lecturas descartadas:", lecturas_descartadas)
print("Promedio PM2.5 (ug/m3):", promedio_pm25)
```

ejercicio 7
— Estadística descriptiva de la serie limpia
Con lecturas_validas y promedio_pm25 del Ejercicio 6, calcule el resto de la descripción de la serie. Necesitará dos recorridos: el primero para los extremos, el segundo para la dispersión (que depende del promedio, y por eso no puede calcularse en la misma pasada).

```python
# 1. Inicialización de los extremos
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

# 2. Primer recorrido: Determinación de máximo y mínimo
for lectura in lecturas_validas:
    if lectura > maximo_pm25:
        maximo_pm25 = lectura
    if lectura < minimo_pm25:
        minimo_pm25 = lectura

# 3. Cálculo del rango físico
rango_pm25 = maximo_pm25 - minimo_pm25

# 4. Segundo recorrido: dispersion
suma_desviaciones = 0.0
for lectura in lecturas_validas:
    suma_desviaciones += (lectura - promedio_pm25) ** 2

# 5. Cálculo de la desviación
desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

# Mostrar resultados
print("Máximo PM2.5:", maximo_pm25)
print("Mínimo PM2.5:", minimo_pm25)
print("Rango PM2.5:", rango_pm25)
print("Desviación estándar:", desviacion_estandar)
```

Ejercicio 8
 — Decaimiento tras cerrar la fuente de emisión
Al detectar el pico de contaminación se cierra la fuente de emisión cercana a la estación. A partir de ese momento, la concentración medida baja un 15 % cada hora. Se quiere saber cuántas horas tardará en volver a categoría "Buena" (12.0 µg/m³ o menos).

```python
# 1.concentración inicial que es el pico del Ejercicio 7 y contadores
concentracion_actual = maximo_pm25
horas_transcurridas = 0

# 2. Bucle que continúa mientras esté por encima de 12.0 ug/m3
while concentracion_actual > 12.0:
    concentracion_actual *= 0.85
    horas_transcurridas += 1

# Mostrar resultados
print("Horas requeridas para llegar a categoría 'Buena':", horas_transcurridas)
print("Concentración final alcanzada (ug/m3):", concentracion_actual)
```
